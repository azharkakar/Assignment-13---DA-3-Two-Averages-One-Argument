```python

df = pd.read_csv(r"C:\Users\LENOVO\Downloads\teaching_prices_messy.csv")
```

```python
#problem 1: casing error
df["item"] = df["item"].str.upper()
```

```python
# Problem 2: remove commas from numbers and convert to numeric
df["price"] = (df["price"].astype(str).str.replace(",", "", regex=False))
```

```python
# Problem 3: drop duplicate rows
df = df.drop_duplicates()
```

```python
# Problem 4: flag missing values
df["missing_flag"] = df.isna().any(axis=1)
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
